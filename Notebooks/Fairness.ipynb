```python
import pandas as pd
from functools import partial
from fairlearn.metrics import demographic_parity_difference, equalized_odds_difference
from fairlearn.metrics import MetricFrame
from sklearn.metrics import recall_score

recall_bad = partial(recall_score, pos_label=2)

def fairness_report(df, label=""):
    y_true = df["target"]
    y_pred = df["prediction_label"]
    sensitive = df["age_group"]

    # Remap y_true and y_pred from (1, 2) to (0, 1) for fairlearn metrics
    y_true_binary = y_true.map({1: 0, 2: 1})
    y_pred_binary = y_pred.map({1: 0, 2: 1})

    dpd  = demographic_parity_difference(y_true_binary, y_pred_binary, sensitive_features=sensitive)
    eod  = equalized_odds_difference(y_true_binary, y_pred_binary, sensitive_features=sensitive)
    mf   = MetricFrame(metrics=recall_bad, y_true=y_true, y_pred=y_pred, sensitive_features=sensitive)

    print(f"[{label}]")
    print(f"  DPD:  {dpd:.4f}")
    print(f"  EOD:  {eod:.4f}")
    print()
    return {"config": label, "DPD": round(dpd, 4), "EOD": round(eod, 4)}


configs = {
    'trtr_rf' : pd.read_csv('/content/drive/MyDrive/Thesis/Datasets/trtr-rf.csv'),
    'trtr_lda' : pd.read_csv('/content/drive/MyDrive/Thesis/Datasets/trtr-lda.csv'),
    'trtr_dummy' : pd.read_csv('/content/drive/MyDrive/Thesis/Datasets/trtr-dummy.csv'),
    'tstr_16_rf' : pd.read_csv('/content/drive/MyDrive/Thesis/Datasets/tstr16-rf.csv'),
    'tstr_16_lda' : pd.read_csv('/content/drive/MyDrive/Thesis/Datasets/tstr16-lda.csv'),
    'tstr_16_dummy' : pd.read_csv('/content/drive/MyDrive/Thesis/Datasets/tstr16-dummy.csv'),
    'tstr_17_rf' : pd.read_csv('/content/drive/MyDrive/Thesis/Datasets/tstr17-rf.csv'),
    'tstr_17_lda' : pd.read_csv('/content/drive/MyDrive/Thesis/Datasets/tstr17-lda.csv'),
    'tstr_17_dummy' : pd.read_csv('/content/drive/MyDrive/Thesis/Datasets/tstr17-dummy.csv'),
}


# Run fairness reports
results = []
for config, config_path in configs.items():
    df = config_path
    df["age_group"] = (df["age"] < 25).map({True: "young (<25)", False: "adult (≥25)"})
    results.append(fairness_report(df, config))

# Save summary
fairness_df = pd.DataFrame(results)
print(fairness_df.to_string(index=False))
fairness_df.to_csv('/content/drive/MyDrive/Thesis/Datasets/fairness_results_age.csv', index=False)
```
